```python
from transformers import pipeline
```

```python
generator = pipeline(task = "text-generation",
                     model = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
                     )
```

```python
print(type(generator))
print(generator)
```

```python
messages = (
    [
        {
            "role" : "user",
            "content" : "Tell me about Pakistan."
        }
    ]
)

response = generator(messages,
                     max_new_tokens = 100, # decides how much the model will genertes
                     do_sample = False)    # controls the modle should be creative or deterministic , if True, it introduces randomness.
```

```python
messages = []
while True:
  user_input = input("You: ")

  if user_input.lower() == "exit":
    break

# store user input
  messages.append(
      {
          "role" : "user",
          "content" : user_input
      }
  )

  # store assitant reply

  response = generator(
      messages,
      max_new_tokens = 100,
      do_sample = True,
      temperature = 0.7
  )

  assistant_reply = response[0]["generated_text"][-1]["content"]
  print("Assistant Reply: " , assistant_reply)
  messages.append(
      {
          "role" : "assistant",
          "content" : assistant_reply
      }
  )

print(messages)
```
